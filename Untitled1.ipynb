```python
STATE_CODES = {
    'ap','ar','as','br','cg','ga','gj','hr','hp','jh','jk','ka','kl',
    'mp','mh','mn','ml','mz','nl','od','pb','rj','sk','tn','tg','tr',
    'up','ut','wb','ch','dd','dl','ld','py','an'
}

def extract_state_block(addr):
    toks = normalize_latin(addr).split()

    for t in reversed(toks[-3:]):
        if t in STATE_CODES:
            return t

    return ''

chunk['block_state'] = (
    chunk['country'].fillna('').astype(str).str.lower() + '|' +
    chunk['business_address'].fillna('').apply(extract_state_block)
)
```

```python
def load_source_to_sqlite(path, table):
    first = True

    for chunk in pd.read_csv(
        path,
        sep='\t',
        dtype=str,
        chunksize=20000
    ):
        # Fill missing values
        chunk['country'] = (
            chunk['country']
            .fillna('')
            .str.lower()
            .str.strip()
        )

        chunk['business_name'] = chunk['business_name'].fillna('')
        chunk['business_address'] = chunk['business_address'].fillna('')

        # -------------------------
        # Normalize business name
        # -------------------------
        names = chunk['business_name'].apply(normalize_name)

        chunk['norm_name_lat'] = [n[0] for n in names]
        chunk['norm_name_trans'] = [n[1] for n in names]
        chunk['norm_name'] = chunk['norm_name_trans']

        # -------------------------
        # Normalize address
        # -------------------------
        chunk['norm_addr'] = (
            chunk['business_address']
            .apply(normalize_latin)
        )

        # -------------------------
        # Name tokens
        # -------------------------
        toks = chunk['norm_name'].apply(lambda x: x.split())

        chunk['first_tok'] = toks.apply(
            lambda x: x[0] if x else ''
        )

        chunk['last_tok'] = toks.apply(
            lambda x: x[-1] if x else ''
        )

        # -------------------------
        # Blocking keys
        # -------------------------
        chunk['block1'] = (
            chunk['country'] + '|' +
            chunk['norm_name'].str[:4]
        )

        chunk['block2'] = (
            chunk['country'] + '|' +
            chunk['first_tok']
        )

        chunk['block3'] = (
            chunk['country'] + '|' +
            chunk['last_tok']
        )

        chunk['block_state'] = (
            chunk['country'] + '|' +
            chunk['business_address'].apply(
                extract_state_block
            )
        )

        # -------------------------
        # Write to SQLite
        # -------------------------
        chunk.to_sql(
            table,
            conn,
            if_exists='append',
            index=False
        )

        # Create indexes only after first chunk
        if first:
            for col in [
                'block1',
                'block2',
                'block3',
                'block_state',
                'country'
            ]:
                conn.execute(
                    f"CREATE INDEX idx_{table}_{col} "
                    f"ON {table}({col})"
                )

            conn.commit()
            first = False

        del chunk
        gc.collect()

    print(f"Finished loading {table}")
```
